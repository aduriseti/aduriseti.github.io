# This is my brain on Julia...
A few months back my company prototyped one of our bidding projects in Julia to try and address performance issues with the current python implementation.  That project didnt end up panning out, but what I saw of Julia impressed me.  A staticly typed, compiled language, with the rich type inference and dead simple syntax of python?  Native support for Nd matrices - speed comparable to numpy or c/++?  Nice!

## Pros
So far, I've been really happy working with Julia, especially compared to any other staticly typed, compiled language I worked with.  Setup was a breeze - and the vscode extension is amazing.  Julia works seamlessly with vscode's jupyter tooling and the linting worked out of the box for me.  Except for Python, Julia has easily has the best Jupyter support of any language I've worked with.  Error messages are clear and to the point, the package manager is great, and the syntax is super simple and concise, comparable to python or go.

I think what I like most about Julia is that I really think I can do anything with it.  If I want to treat it as a scripting language like js or python, I can quickly type out the functionality I want.  If I want to use it for scientific computing, the ecosystem is there and comparable to something like Matlab (w/o the exorbitant cost).  But, when it comes time to write more serious production code, I have the tools to optimize it and (importantly) get speedups from multithreading. 

## Cons
Unfortunately Julia also has its flaws.  Its primarily a scientific computing language - AFAIK there are no large companies who use Julia as a core part of their stack.  

Syntactically speaking - the 1-based indexing is a bizarre choice and I find myself subtracting 1 from many index vars just to emulate 0-indexing - but given Julia is a scientific language it kindof makes sense that it would inherit some traits from Matlab.  While I enjoy being able to define blocks and capture their end value w/ `begin...end` and being able to control scope w/ `let...end`,  I dont like having type out multi character keywords all the time.  I also think they interfere w/ the readability of the code if I follow patterns Im used to in javascript or c++.  I miss the `i++` and `++i` operators - the equivalent block to `i++` in julia is basically unusable: `begin tmp = i; i += 1; tmp end`.  

I also find the OO elements of Julia a little clunky.  Sometimes I want objects w/ member methods that have access to data scoped to that object.  You can techincally use the JS function == object == class approach in Julia - but its clunky, the syntax is unintuitive, and linting doesnt work.

````julia
Heap = () -> begin
    size = 0
    li = (n) -> 2*n + 0
    ri = (n) -> 2*n + 1
    pi = (n) -> floor(Int,n/2)
    return () -> (size;li;ri;pi)
end
o = Heap()
("o.size",o.size,"o.li(1)",o.li(1),"o.ri(1)",o.ri(1),"o.pi(1)",o.pi(1))
````

You can do something similar w/ classes having functions as member variables, but again, linting doesnt work and I get the feeling this just isnt how you are supposed to write Julia.

## Verdict
A lot of companies use python for data engineering and datascience, and a more performant language like Java or Go for their backend.  Or they might use python for ML/datascience, and Java for their pipeline and backend.  Some companies even use python or JS/TS for their backend services, trading performance for ease-of-use.  

I think Julia has the potential to satisfy all backend requirements w/o the need for tradeoffs.  AFAICT it has [good performance](https://betterprogramming.pub/why-you-should-invest-in-julia-now-as-a-data-scientist-30dc346d62e4) and good support for threading.  I see a lot of the packages and connectors you might need for data & app engineeering (DB, ORM, kafka, spark, AWS sdk, etc...) - and Julia of course satisfies any datascience or ML need.  

BUT - Julia just isnt as common as the industry standards and doesnt have the backing of any large company.  Coming from python, comparable packages in Julia arent as mature as you might like (this was the main reason my company decided not continue down the Julia path).

Overall?  Id like to use Julia, and will continue to advocate it for backend tasks when possible - but I just dont see any manager agreeing.  Bummer :/

## RSA
can find positive integers `e`,`d`,`n` s.t
```
(m^e)^d \congruent(n) m

(m^d)^e \congruent(n) m
```
public key is then `n` and `e` - private key is `n` and `d` - although only `d` is actually kept private

`m` is the encrpyed payload


## Key gneeration
1. chose prime #s `p` `q` - similarly sized ,larged, random - differ in len by a few digits
2. `n = pq` - bits in n is the key length - as in rsa2556
3. carmichael fn `\lam`, `lam(n) = m smallest s.t. a^m % n = 1 \forall a coprime to n`
    - `\lam(n)` = `lcm(lam(p),lam(q))` \
        `n = pq` => `a ^ \lam(n) % n = 1` \
        => `a ^ \lam(p) = 1 % pq`  && `a ^ \lam(q) = 1 % pq` \
        ```
        1/pq = 1/p 
        (x / (pq)) % pq = x/p % pq * x/q % pq
        a ^ lcm(\lam(p),\lam(q)) % (pq) = a ^ s ^ \lam(p) % (pq) = a ^ t ^ \lam(q) % (pq) = 1
        a ^ s ^ \lam(p) / p * 1/q = (k + 1/p) * 1/q = (pk + 1)/pq
                                                    = (lq + 1)/pq
        pk = lq => divisibnle by both p and q
                                                    => % pq = 1
        any multiple of \lam(q) and \lam(p) will be congruent to 1 but lcm is lowest such
        ```
    - for all primes carmichaels totient \lam = eulers toitent \phi
    - for all primes eulers toitent = `p-1`
    - also Fermats little theorem
        ```
        a^p \congruent(p) a
        a^(p-1) \congruent(p) 1
        ```
        - Proof:
            - consider dictionary of strings of length `p` w/ alphabet of size `a`
            - remove all strings completely composed of the same character - which is `a`
            - all remaining strings may be grouped by whether they are equal if ends are connected
            - these groups are of size p
            - so `(a^p - a) % p = 0` => `a^p /congruent(p) a`
            - can divide both sides by `a` => `a^(p-1) % p = 1`
                - `a^p % p = a % p`
            - what if `p|a`? - in this case `a^(p-1) % p = 0`
            - what if `a|p`? a = 2 , p = 4
                `2 ^ (4-1) % 4 = 0`
            - b/c common divisor of a,p is divisible by p when exponentiated enough
    - `\lam(n) = lcm(\lam(p),\lam(q)) = lcm(p-1,q-1)`
    - `lcm` computed used Euclidean algo - 
        - `lcm(m,n) = m * n / gcd(m,n)`
4. chose `e` s.t `1 < e < \lam(n)` & `gcd(e,\lam(n)) = 1` - i.e. coprime `e` and `lam(n)`
    - `e` should have a small bit length and hamming weight `3` smalest and fastest - but `2^16 + 1` is most common
    - `e` part of public key
5. `d` is then `d %=\lam(n) e^-1` - 
    => `de % \lam(n) = 1`
    - d computed via Extended Euclidean Algo b/c `e` & `\lam(n)` comprime
    - `d` kept secret as private key exp

public key is then modulus `n` and public encryption exponent `e` - `d` is then the private decrpytion exponeent 

`p`, `q` and `\lam(n)` can be kepy secret b/c they can be used to calcujlate `d` - but dont need to be remembered

***NOTE***: apparently eulers totient can be used instead of carmichaels - so 
            `(p-1)(q-1)` works just as well as `\lcm(p-1,q-1)`
            - but this makes decription key larger than it need to be 

***NOTE***: decryption key can be chosed before encryption key to make decryption eky smaller and faster

## Encryption
message `m`

encrypted payload = `m ^ e % n`

## Decryption
payload 'p = m^e % n`

`p^d % n = m^e^d % n = m`

***NOTE*** Chinese remainder theorem used for decryption

## Goal
Modular exponentiation is irreversable "1-way fn" - so if we can find `e`,`d`,`n` s.t.
for any message integer `m`:
```
(m^e)^d = m % n
```
then we can encrypt `m` via modular exponentiation over `n` by `e` - `n` and `e` then form 
public (encryption) key

we can then decrypt this encrpyted message by modular exponentiation over `n` by `d`
`n` and `d` then form private (decryption) key - although only `d` is kept private 

## Key generation
1. choose 2 large primes `p`,`q` - differ in length by a few digits - random - large
2. `n = pq`
3. Want to find a `k` s.t for all `m` =====>>> `m^k = 1 % n`
    - FLT:
        - for prime `n` - `a^n - a = 0 mod n` - for all `a`
            - alphabet of size `a`
            - universe of strings of length `m`
        => `a^m = a` && `a^(m - 1) = 1 mod n` && `a^(n-2) = 1/a mod n`
    ```
    m^k = 1 % (pq)
    By FLT:
    m^(p-1) = 1 % p && m^(q-1) = 1 
    m^(p-1) % pq = (s*p + 1) % pq
    m^(q-1) % pq = (t*q + 1) % pq
    m^(p-1)(q-1) = (t*q + 1) % pq
                 = (p*s + 1) % pq
                 = (k*p*q / gcd(pq) + 1) % pq
                 = 1
    ```
    So we have `k = (p-1)(q-1)` AND `m^(p-1)(q-1) = 1 % pq`
    NOTE - cant do `m^((p-1)*(q-1)+1) % pq` b/c that requires u to know `m` at decryption time!
4. WTF `e,d` s.t. `e*d = i*k + 1` where `i` is some int
            => `e*d = 1 % k`
    If we choose `e` prime < `k` ==>> `gcd(e,k) = 1` - then last step of Euclids `gcd(e,k)` is 
    ```
    ae + bk = gcd(e,k) = 1
    ```
    so `a` = `d`

    Can also use FLT b/c `e`,`k` coprime
    ```
    e^(k-1) = 1 % k
     = e * e^(k-2) = 1 % k
    d = e ^ (k-2) % k
    ```

x1*a + y1*b = gcd(a,b)
